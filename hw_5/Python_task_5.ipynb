# **POLARS**

**1) Считывание датасета**

```python
import polars as pl

df = pl.read_csv('train.csv')
```

```python
print('Статистическое описание датасета train.csv\nСредние значения указаны в строке "mean"')

df.describe()
```

**3. Подсчет количества пассажиров каждого класса**

```python
print(df.get_column("Pclass").value_counts().sort("Pclass"))
```

**4. Вывод количества выживших мужчин и женщин на корабле**

```python
print("Количество выживших по полу")
print(df.group_by(['Sex']).agg(pl.col('Survived').sum()))

```

**5. Вывод части таблицы, где возраст пассажиров больше 44 лет**

```python
print(df.filter(pl.col('Age') > 44))
```

# **УСКОРЕНИЕ РАБОТЫ С PANDAS**

**1. Считывание датасета** 

```python
import pandas as pd
import bottleneck as bn

df_pd = pd.read_csv('train.csv')
```

**2. Подсчет среднего возраста пассажиров и его стандартного отклонения с помощью bottleneck**

```python
print(f"Средний возраст пассажиров: {bn.nanmean(df_pd['Age']):.0f}")
print(f"Стандартное отклонение возраста пассажиров: {bn.nanstd(df_pd['Age']):.2f}")
```

**3. Получение столбца Fare_new как Fare * 1.3**

```python
df_pd['Fare_new'] = df_pd['Fare'].apply(lambda x: x * 1.3)
print(df_pd[['Fare', 'Fare_new']])
```

# **ОПТИМИЗАЦИЯ ТИПОВ PANDAS**

**1. Считывание датасета Housing.csv**

```python
df_house = pd.read_csv("Housing.csv")
```

**2. Определение оптимального типа данных для колонок датасета**

```python
df_house.info(memory_usage='deep') 
```

Мы видим 2 основных типа данных: int64 и object. Для числовых значений мы можем уменьшить **разрядность**. Кроме того можно учесть, являются ли числа неотрицательными. Для object мы можем воспользоваться **категориальным** типом данных

```python
print("\nДиапазоны числовых столбцов:")
numeric_data = ['price', 'area', 'bedrooms', 'bathrooms', 'stories', 'parking']
for col in numeric_data:
    print(f"{col}: min = {df_house[col].min()}, max = {df_house[col].max()}")
```

Соответственно для **price** - **uint32** (от 0 до 4.294.967.295), **area** - **uint16** (от 0 до 65535), для **bedrooms**, **bathrooms**, **stories**, **parking** - **uint8** (от 0 до 255) 

**3. Смена типа данных столбцов датасета**

```python
categorical_data = ['mainroad', 'guestroom', 'basement', 'hotwaterheating', 'airconditioning', 'prefarea', 'furnishingstatus']
opt_numeric = ["uint32", "uint16", "uint8", "uint8", "uint8", "uint8"]
for col, type in zip(numeric_data, opt_numeric):
    df_house[col] = df_house[col].astype(type)
for col in categorical_data:
    df_house[col] = df_house[col].astype('category')
```

```python
df_house.info(memory_usage='deep') 
```
